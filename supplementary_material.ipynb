# Libraries

```python
import pandas as pd
from IPython.display import Image
import matplotlib.pyplot as plt
from IPython.display import Video
```

# Functions

```python
def figure(path):

    img = plt.imread(path)

    fig, ax = plt.subplots(figsize = (12,10))
    ax.imshow(img)
    ax.set_axis_off()
#----------------------------------------------------------------------------------------
def photographs(path1, path2, path3, path4):

    photo1 = plt.imread(photo_1)
    photo2 = plt.imread(photo_2)
    photo3 = plt.imread(photo_3)
    photo4 = plt.imread(photo_4)
    
    fig, ax = plt.subplots(2, 2, figsize = (10,8))
    ax[0,0].annotate(xy = (3500, 250), text = 'A', color = 'red')
    ax[0,0].imshow(photo1, aspect = 'auto')
    ax[0,0].set_axis_off()
    ax[0,1].annotate(xy = (3500, 250), text = 'B', color = 'red')
    ax[0,1].imshow(photo2, aspect = 'auto')
    ax[0,1].set_axis_off()
    ax[1,0].annotate(xy = (3500, 250), text = 'C', color = 'red')
    ax[1,0].imshow(photo3, aspect = 'auto')
    ax[1,0].set_axis_off()
    ax[1,1].annotate(xy = (3500, 250), text = 'D', color = 'yellow')
    ax[1,1].plot([1600, 2900], [1000, 900], color = 'red', linewidth = 2.5)
    ax[1,1].imshow(photo4, aspect = 'auto')
    ax[1,1].set_axis_off();    
#----------------------------------------------------------------------------------------
def figure2(path):

    img = plt.imread(path)

    fig, ax = plt.subplots(figsize = (8,16))
    ax.imshow(img)
    ax.set_axis_off()
#----------------------------------------------------------------------------------------
def dataframe(path1, path2):

    dataframe1 = pd.DataFrame(pd.read_excel(path1)).style.set_caption("Dataframe A")
    
    dataframe1.format(precision=2, thousands=",", decimal=".")

    dataframe2 = pd.DataFrame(pd.read_excel(path2)).style.set_caption("Dataframe B")
    
    dataframe2.format(precision=2, thousands=",", decimal=".")

    return dataframe1, dataframe2
#----------------------------------------------------------------------------------------
def figure_jpeg(path):
    
    return Image(filename=path, width=800, height=600)
#----------------------------------------------------------------------------------------
def figure_jpeg2(path):
    
    return Image(filename=path, width=600, height=400)
```
